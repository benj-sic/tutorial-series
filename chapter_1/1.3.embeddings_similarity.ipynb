# 1.3 Embedding and Similarity

Section 1.2 showed how text becomes tokens and how the context window limits what an LLM can receive. What happens after those tokens enter the model?

Computers cannot directly compute operations on strings like "kinase" or "inhibitor." Mapping tokens to high-dimensional numeric vectors called **embeddings** places text in a continuous numerical vector space. This translates semantic relationships into spatial coordinates and allows us to perform geometric and linear algebra operations to measure conceptual similarity mathematically.

**Weights** are the trainable numeric matrices within the neural network's layers that dictacte how input vectors, interact, transform, and pass signals to subequent transformer layers. While an initial embedding maps a token to a baseline vector, the model's self-attention weights update the vector based on neighboring words (e.g., distinguishing a cell "culture" vs. a lab "culture").

**Training** is how weights are learned. The model processes massive amounts of text (e.g., PubMed, arXiv, etc.) and attempts to predict next tokens. The model compares its prediction to the actual text, calculates the error, and adjusts its weights iteratively via backpropagation.

Prompt tokens are converted into vectors and passed through the model's layers. Using its learned weights, the output layer produces a probability distribution over the entire vocabularly or what token should come next. The model selects the most likely token, appends it to the context, and runs another forward pass to predict the token after that.

Tokens are converted into numerical representations. During a forward pass, the model uses its learned weights and the surrounding context to calculate a prediction for the next token. It generate a response by repeating this process. Training is how the weight were learned; supplying a new prompt uses those weights rather than retraining the model.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

text = "cell culture"

embedding = model.encode(text)

n = 50

print("Text:", text)
print("Embedding size:", embedding.shape[0])
print("Embedding:", embedding[:50], f"({n} out of {embedding.shape[0]})")
```

```python
texts = [
    "cell culture",
    "growing cells in a lab",
    "lab culture"
]

embeddings = model.encode(texts)
similarities = model.similarity(embeddings, embeddings)

print(f"{texts[0]} vs. {texts[1]}: {similarities[0, 1]:.3f}")
print(f"{texts[0]} vs. {texts[2]}: {similarities[0, 2]:.3f}")
print(f"{texts[1]} vs. {texts[2]}: {similarities[1, 2]:.3f}")
```
